```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40])
print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 2)
```

```python
arr = np.array([10, 20, 30, 40])
print(np.sum(arr))
```

```python
arr = np.array([10, 20, 30, 40])
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.std(arr))
```

```python
arr = np.array([-10, -20, 30, -40])
print(np.abs(arr))
```

```python
arr = np.array([1, 4, 9, 16])
print(np.sqrt(arr))
print(np.exp(arr))
print(np.log(arr))
```

```python
arr = np.array([1.234, 2.567, 3.891])
print(np.round(arr))
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(np.sum(arr))
print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
```
