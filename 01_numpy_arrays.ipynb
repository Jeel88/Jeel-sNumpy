**NumPy is designed for numerical operations on entire arrays.**<br>
*So every operation are done on arrays only*

```python
import numpy as np #Importing Library
```

```python
numbers=np.array([10,20,30,40,50])  #Defining Arrays List 1D
print(numbers) #Printing Numbers
```

```python
Numbers2D=np.array([[10,20,30],[40,50,60]]) #2D Array List Brackets Uses
print(Numbers2D) #Printing Numbers 2D
```

```python
Numbers3D=np.array([[[10,20,30],[40,50,60]],[[70,80,90],[110,120,130]]]) #3D Array List Brackets Uses
print(Numbers3D) #Printing Numbers 2D
```
