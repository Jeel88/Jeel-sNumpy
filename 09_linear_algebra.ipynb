```python
import numpy as np
```

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A.T)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
print(A @ B)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))
```
